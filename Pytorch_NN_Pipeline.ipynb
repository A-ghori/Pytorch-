# PyTorch: Autograd, Computation Graph, Leaf Tensors & Gradient Descent

## 1. What is Autograd?

**Autograd (Automatic Differentiation)** is PyTorch's automatic differentiation engine.

It automatically calculates gradients (derivatives) of tensors with respect to other tensors.

In Machine Learning, gradients help us understand how much a parameter (weight or bias) affects the loss.

### Mathematical Example

Suppose:

$$
y=x^2
$$

Differentiating with respect to \(x\):

$$
\frac{dy}{dx}=2x
$$

If:

$$
x=3
$$

Then:

$$
\frac{dy}{dx}=2(3)
$$

$$
\boxed{\frac{dy}{dx}=6}
$$

PyTorch can calculate this derivative automatically using Autograd.

### Python Implementation

```python
import torch

x = torch.tensor(3.0, requires_grad=True)

y = x ** 2

y.backward()

print(x.grad)
```

**Output:**

```text
tensor(6.)
```

### Explanation

1. `torch.tensor(3.0, requires_grad=True)` creates a tensor and enables gradient tracking.
2. `y = x ** 2` performs a mathematical operation.
3. PyTorch tracks this operation.
4. `y.backward()` calculates the derivative of `y` with respect to `x`.
5. `x.grad` stores the calculated gradient.

**Important:** `requires_grad=True` tells PyTorch to track operations involving this tensor so that gradients can be calculated later.

## 2. What is a Computation Graph?

A computation graph represents the sequence of mathematical operations performed on tensors.

Consider:

```python
x = torch.tensor(3.0, requires_grad=True)

y = x ** 2

z = y + 5
```

### Mathematical Representation

$$
y=x^2
$$

$$
z=y+5
$$

Substituting the value of \(y\):

$$
z=x^2+5
$$

If:

$$
x=3
$$

Then:

$$
z=3^2+5
$$

$$
\boxed{z=14}
$$

### Computation Graph

```text
       x = 3
          |
          v
       y = x²
          |
          v
       z = y + 5
          |
          v
        z = 14
```

PyTorch tracks these operations so that it can calculate gradients during the backward pass.

### Calculate the Gradient

```python
z.backward()

print(x.grad)
```

Mathematically:

$$
z=x^2+5
$$

Differentiating with respect to \(x\):

$$
\frac{dz}{dx}=2x
$$

Substituting \(x=3\):

$$
\frac{dz}{dx}=2(3)
$$

$$
\boxed{\frac{dz}{dx}=6}
$$

**Output:**

```text
tensor(6.)
```

## 3. What is a Leaf Tensor?

A leaf tensor is a tensor at the starting point of a computation graph.

In the usual case, a tensor directly created by the user with `requires_grad=True` is a leaf tensor.

### Example

```python
x = torch.tensor(5.0, requires_grad=True)

y = x * 2

z = y + 10
```

### Understanding the Tensors

| Tensor | Value | Role |
|---|---:|---|
| `x` | 5 | Leaf tensor |
| `y` | 10 | Intermediate tensor |
| `z` | 20 | Intermediate tensor |

### Why?

- `x` was directly created by us.
- `y` was calculated using `x`.
- `z` was calculated using `y`.

Therefore, `x` is the leaf tensor.

### Verify Using Python

```python
print(x.is_leaf)
print(y.is_leaf)
print(z.is_leaf)
```

**Output:**

```text
True
False
False
```

### Why Are Weights and Bias Leaf Tensors?

In our neural network:

```python
self.weights = torch.rand(
    X.shape[1],
    1,
    dtype=torch.float64,
    requires_grad=True
)

self.bias = torch.zeros(
    1,
    dtype=torch.float64,
    requires_grad=True
)
```

Both weights and bias are directly initialized by us.

Therefore, they are leaf tensors.

They are also model parameters that we want to optimize during training.

## 4. What is Backpropagation?

Backpropagation is the process of calculating gradients of the loss with respect to model parameters.

Suppose our neural network is:

### Step 1: Linear Transformation

$$
z=XW+b
$$

Where:

- \(X\) = Input features
- \(W\) = Weights
- \(b\) = Bias

### Step 2: Activation Function

$$
\hat{y}=\sigma(z)
$$

The sigmoid function is:

$$
\sigma(z)=\frac{1}{1+e^{-z}}
$$

It converts the output into a probability between 0 and 1.

### Step 3: Calculate Loss

For binary classification, we use Binary Cross Entropy:

$$
L=-\frac{1}{N}\sum_{i=1}^{N}
\left[
y_i\log(\hat{y}_i)
+
(1-y_i)\log(1-\hat{y}_i)
\right]
$$

Where:

- \(N\) = Number of training samples
- \(y_i\) = Actual label
- \(\hat{y}_i\) = Predicted probability

### Step 4: Calculate Gradients

Our objective is to minimize the loss.

For that, we calculate:

$$
\frac{\partial L}{\partial W}
$$

And:

$$
\frac{\partial L}{\partial b}
$$

These derivatives tell us how the loss changes when weights or bias change.

PyTorch performs this calculation using:

```python
loss.backward()
```

After backward propagation:

```python
print(model.weights.grad)

print(model.bias.grad)
```

The `.grad` attribute stores the calculated gradients.

---

## 5. What is Gradient Descent?

Gradient Descent is an optimization algorithm used to minimize the loss function.

It updates model parameters in the opposite direction of their gradients.

### Mathematical Formula

For weights:

$$
W_{\text{new}}
=
W_{\text{old}}
-
\alpha\frac{\partial L}{\partial W}
$$

For bias:

$$
b_{\text{new}}
=
b_{\text{old}}
-
\alpha\frac{\partial L}{\partial b}
$$

Where:

- \(W_{\text{old}}\) = Previous weight
- \(W_{\text{new}}\) = Updated weight
- \(b_{\text{old}}\) = Previous bias
- \(b_{\text{new}}\) = Updated bias
- \(\alpha\) = Learning rate
- Gradient = Derivative of loss with respect to a parameter

### Numeric Example: Weight Update

Suppose:

$$
W_{\text{old}}=2.0
$$

$$
\text{Gradient}=4.0
$$

$$
\alpha=0.1
$$

Using the gradient descent formula:

$$
W_{\text{new}}
=
W_{\text{old}}-\alpha\times\text{Gradient}
$$

Substituting the values:

$$
W_{\text{new}}=2.0-(0.1\times4.0)
$$

$$
W_{\text{new}}=2.0-0.4
$$

$$
\boxed{W_{\text{new}}=1.6}
$$

The weight changes from \(2.0\) to \(1.6\).

### Numeric Example: Bias Update

Suppose:

$$
b_{\text{old}}=0
$$

$$
\text{Gradient}=0.5
$$

$$
\alpha=0.1
$$

Using:

$$
b_{\text{new}}
=
b_{\text{old}}-\alpha\times\text{Gradient}
$$

Substituting:

$$
b_{\text{new}}=0-(0.1\times0.5)
$$

$$
b_{\text{new}}=0-0.05
$$

$$
\boxed{b_{\text{new}}=-0.05}
$$

The bias changes from \(0\) to \(-0.05\).

**Important:** These are illustrative values. Actual gradients are calculated by PyTorch from the model's loss.

## 6. What is torch.no_grad() and Why Do We Need It?

Consider our parameter update:

```python
model.weights -= learning_rate * model.weights.grad

model.bias -= learning_rate * model.bias.grad
```

This can produce the following error:

```text
RuntimeError:
a leaf Variable that requires grad is being used
in an in-place operation.
```

### Why Does This Happen?

Our weights and bias are:

- Leaf tensors.
- Created with `requires_grad=True`.
- Tracked by Autograd.

The `-=` operator performs an in-place operation.

An in-place operation modifies the existing tensor directly.

PyTorch does not allow this kind of direct in-place modification of a leaf tensor that requires gradients while gradient tracking is enabled.

### Solution: torch.no_grad()

```python
with torch.no_grad():

    model.weights -= learning_rate * model.weights.grad

    model.bias -= learning_rate * model.bias.grad
```

`torch.no_grad()` temporarily disables gradient tracking inside its block.

### Why Do We Disable Gradient Tracking During Parameter Updates?

There are two different operations during training.

**Operation 1: Calculate Gradients**

```python
y_pred = model.forward(X_train_tensor)

loss = model.loss(y_pred, y_train_tensor)

loss.backward()
```

Here, Autograd must be ON because we need gradients.

**Operation 2: Update Parameters**

```python
with torch.no_grad():

    model.weights -= learning_rate * model.weights.grad

    model.bias -= learning_rate * model.bias.grad
```

Here, we already have the gradients.

We only need to update the parameters. We do not need to calculate gradients of the update operation itself.

Therefore, we temporarily turn off gradient tracking.

**Important:** `torch.no_grad()` does NOT permanently disable `requires_grad`. Once we exit the block, normal gradient tracking resumes.

---

## 7. What is Gradient Accumulation?

PyTorch accumulates gradients by default.

This means every time we call:

```python
loss.backward()
```

New gradients are added to the existing `.grad` values instead of replacing them.

### Numeric Example

Suppose the first backward pass calculates:

$$
G_1=3
$$

After:

```python
loss.backward()
```

The stored gradient becomes:

$$
W.grad=3
$$

Suppose the next backward pass calculates:

$$
G_2=2
$$

After another `loss.backward()`:

$$
W.grad=G_1+G_2
$$

$$
W.grad=3+2
$$

$$
\boxed{W.grad=5}
$$

But during normal training, we generally want to calculate fresh gradients for each iteration.

Therefore, we reset them:

```python
model.weights.grad.zero_()

model.bias.grad.zero_()
```

The `.zero_()` method sets the existing gradient values to zero.

**Note:** This accumulation behavior is important when using multiple backward passes, gradient accumulation over mini-batches, or custom training loops.

## 8. Understanding the Complete Training Loop

Now we can combine all the concepts.

### Python Implementation

```python
for epoch in range(epochs):

    # 1. Forward Pass
    y_pred = model.forward(X_train_tensor)

    # 2. Calculate Loss
    loss = model.loss(y_pred, y_train_tensor)

    # 3. Backward Pass
    loss.backward()

    # 4. Update Parameters
    with torch.no_grad():

        model.weights -= learning_rate * model.weights.grad

        model.bias -= learning_rate * model.bias.grad

    # 5. Reset Gradients
    model.weights.grad.zero_()

    model.bias.grad.zero_()

    print(f"Epoch: {epoch}")
    print(f"Loss: {loss.item()}")
    print(f"Bias: {model.bias.item()}")
```

### Understanding the Training Cycle

```text
          START EPOCH
               |
               v
          FORWARD PASS
               |
               v
            PREDICTION
               |
               v
          CALCULATE LOSS
               |
               v
          BACKWARD PASS
          loss.backward()
               |
               v
        CALCULATE GRADIENTS
          weights.grad
           bias.grad
               |
               v
        UPDATE PARAMETERS
         (torch.no_grad)
               |
               v
         RESET GRADIENTS
             zero_()
               |
               v
           NEXT EPOCH
```

## 9. Understanding Why Bias Changes From Zero

Initially, we initialize bias as:

```python
self.bias = torch.zeros(
    1,
    dtype=torch.float64,
    requires_grad=True
)
```

Initial value:

$$
b_0=0
$$

During training, the gradient is calculated:

$$
\frac{\partial L}{\partial b}
$$

Then the bias is updated:

$$
b_{\text{new}}
=
b_{\text{old}}
-
\alpha\frac{\partial L}{\partial b}
$$

Suppose:

$$
b_{\text{old}}=0
$$

$$
\alpha=0.1
$$

$$
\frac{\partial L}{\partial b}=0.5
$$

Then:

$$
b_{\text{new}}=0-(0.1\times0.5)
$$

$$
b_{\text{new}}=-0.05
$$

After multiple epochs, the bias may become negative or positive depending on the gradients.

For example:

```python
print(model.bias)
```

Output:

```text
tensor([-0.9296], dtype=torch.float64, requires_grad=True)
```

This means the bias has been updated during training.

**Important:** A negative bias does not mean the model is wrong. The model learns whichever parameter values help minimize the loss.

## 10. Important PyTorch Functions: Quick Revision

| Function / Attribute | Purpose |
|---|---|
| `requires_grad=True` | Enables gradient tracking for a tensor |
| `.backward()` | Calculates gradients |
| `.grad` | Stores calculated gradients |
| `.is_leaf` | Checks whether a tensor is a leaf |
| `torch.no_grad()` | Temporarily disables gradient tracking |
| `.zero_()` | Resets gradient values to zero |
| `.item()` | Extracts a scalar tensor as a Python number |
| `.detach()` | Returns a tensor disconnected from the current computation graph |

### Difference Between no_grad() and detach()

**no_grad():**

```python
with torch.no_grad():
    y = model.forward(X_train_tensor)
```

Temporarily prevents operations inside the block from being tracked by Autograd.

**detach():**

```python
y_detached = y_pred.detach()
```

Creates a tensor disconnected from the current computation graph.

They are related, but they are not interchangeable in every situation.

## 11. Final Understanding

The complete training process is:

1. **Forward Pass:** Generate predictions using weights and bias.
2. **Loss Calculation:** Measure the difference between predictions and actual labels.
3. **Backward Pass:** Autograd calculates gradients using backpropagation.
4. **Gradient Descent:** Use the gradients to update weights and bias.
5. **Gradient Reset:** Clear old gradients before the next iteration.
6. **Repeat:** Continue for multiple epochs until the model learns useful parameters.

### Remember These Two Concepts

**Autograd calculates gradients.**

**Gradient Descent uses those gradients to update model parameters.**

They are two different but connected concepts.


````python
# Defining Model

class MySimpleNN():

  def __init__(self,X):
    self.weights = torch.rand(X.shape[1], 1, dtype=torch.float64, requires_grad=True) # 30 x 1 matrix
    self.bias = torch.zeros(1, dtype=torch.float64, requires_grad=True)
    self.loss_fn = torch.nn.BCELoss()

  def forward(self,X):
    z = torch.matmul(X, self.weights) + self.bias
    y_pred = torch.sigmoid(z)
    return y_pred
  #Calculate loss
  def loss (self, y_pred, y):
    loss = self.loss_fn(y_pred, y)
    return loss
````
