```python
import torch
```

```python
x = torch.tensor(6.7)
y = torch.tensor(0.0)
```

```python
w = torch.tensor(1.0, requires_grad=True)
b = torch.tensor(1.0, requires_grad=True)
```

```python
# Forward Pass
z = w * x + b
z
```

```python
# sigmoid activation
y_pred = torch.sigmoid(z)
y_pred
```

```python
# loss
import torch.nn.functional as F
loss = F.binary_cross_entropy(y_pred, y)
loss
```

```python
loss.backward()
```

```python
print(w.grad)
print(b.grad)
```

# X = [1,2,3]
  #   x1,x2,x3
#now differentiate
# X^2.mean = (x1^2 + x2^2 + x3^2) /3.
# dy/dx1 = 2x1 / 3
# dy/dx2 = 2x2 / 3
# dy/dx3 = 2x3 / 3

# Put the values
# dy / dx1 = (2 * 1) / 3 = 2 /3 = 0.666
# dy / dx2 = (2 * 2) / 3 = 4 / 3 = 1.333
# dy / dx3 = (2 * 3) / 3 = 6 / 3 = 2.00  


```python
X = torch.tensor([1.0,2.0,3.0], requires_grad=True)

y = (X ** 2).mean()
y

y.backward()
```

```python
print(X.grad)
```
