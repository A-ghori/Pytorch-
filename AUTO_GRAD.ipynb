```python
import torch
```

```python
x = torch.tensor(6.7)
y = torch.tensor(0.0)
```

```python
w = torch.tensor(1.0, requires_grad=True)
b = torch.tensor(1.0, requires_grad=True)
```

```python
# Forward Pass
z = w * x + b
z
```

```python
# sigmoid activation
y_pred = torch.sigmoid(z)
y_pred
```

```python
# loss
import torch.nn.functional as F
loss = F.binary_cross_entropy(y_pred, y)
loss
```

```python
loss.backward()
```

```python
print(w.grad)
print(b.grad)
```
